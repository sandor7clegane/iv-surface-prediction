```python
## separating each iv value per row for better access
long_df = pd.melt(
    df,
    id_vars=[
        "datetime",
        "underlying_price"
    ],
    value_vars=iv_cols,
    var_name="option",
    value_name="iv"
)
```

```python
## extracting option type from strike string
long_df["option_type"] = long_df["option"].str[-2:]
```

```python
## extracting strike value from strike string
long_df["strike"] = (
    long_df["option"]
    .str[-7:-2]
    .astype(int)
)
```

```python
## calculatin moneyness for each strike and timestamp to check whether it is ATM or OTM/ITM
long_df["moneyness"] = (
    long_df["strike"]
    /
    long_df["underlying_price"]
)
```

```python
## to get more symmetry around 0
long_df["log_moneyness"] = np.log(
    long_df["moneyness"]
)
```

```python
## distance from ATM
long_df["distance_atm"] = abs(
    long_df["strike"]
    -
    long_df["underlying_price"]
)
```

```python
feature_cols = [
    c
    for c in test_df.columns
    if c not in [
        "datetime",
        "underlying_price"
    ]
]
```

```python
col_meta = {}

for c in feature_cols:

    col_meta[c] = {

        "type":
            "CE" if "CE" in c else "PE",

        "strike":
            int(
                c.replace("NIFTY27JAN26","")
                 .replace("CE","")
                 .replace("PE","")
            )
    }
```

```python
ce_cols = sorted(
    [c for c in feature_cols if "CE" in c],
    key=lambda x: col_meta[x]["strike"]
)
```

```python
pe_cols = sorted(
    [c for c in feature_cols if "PE" in c],
    key=lambda x: col_meta[x]["strike"]
)
```

```python
## SURFACE CURVATURE PLOT ANALYSIS:

# the curvature distribution is heavily concentrated around zero
# most strike locations exhibit very low curvature
# we can say that the IV surface is generally smooth
# large local discontinuities are very less
# neighboring strikes contain substantial information
# interpolation-based reconstruction methods are likely to perform well

## AVG. IV BY MONEYNESS PLOT ANALYSIS:

# avg. IV exhibits a pronounced smile-like structure
# min avg. IV occurs near moneyness ≈ 1
# IV increases as moneyness moves away from ATM
# the right tail increases more sharply than the left tail
# the smile is asymmetric, indicating the presence of skew
# moneyness appears to be one of the strongest explanatory variables for IV
```

```python
# from all the earlier experiments mentioned:
# Strike-based linear interpolation
# Strike-based cubic interpolation
# Hybrid interpolation for larger gaps
# Log-moneyness based interpolation
#
# Best performance was obtained after moving from strike
# space to log-moneyness space.
#
# Final approach:
# Fit smiles in log-moneyness coordinates
# Emphasize ATM observations as mentioned
# Use adaptive polynomial fitting
# Stabilize wing extrapolation
# Apply temporal fallback if needed
```

```python
## final submission
blend09_df = test_df.copy()
```

```python
# fill missing IVs smile by smile
# CE and PE handled separately

for i in range(len(test_df)):

    row = test_df.iloc[i]

    # current spot
    S = row["underlying_price"]

    for opt_cols in [ce_cols, pe_cols]:

        # strike grid
        strikes = np.array(
            [col_meta[c]["strike"] for c in opt_cols],
            dtype=float
        )

        # IV values
        iv = row[opt_cols].values.astype(float)

        # observed and missing points
        obs = ~np.isnan(iv)
        missing = ~obs

        # skip complete smiles
        if missing.sum() == 0:
            continue

        # locations with IV available
        obs_idx = np.where(obs)[0]

        # strike -> log moneyness
        lm = np.log(strikes / S)

        # cubic smile fit
        # lower degree if too few points
        deg = min(3, len(obs_idx) - 1)

        # give ATM more importance
        weights = (
            1.0
            /
            (
                1.0
                +
                (lm[obs_idx]**2) * 2250
            )
        )

        # fit smile
        p = np.polyfit(
            lm[obs_idx],
            iv[obs_idx],
            deg,
            w=weights
        )

        # fill every missing point
        for col_idx in np.where(missing)[0]:

            # smile prediction
            pred_poly = np.polyval(
                p,
                lm[col_idx]
            )

            # nearest observed points
            left = obs_idx[
                obs_idx < col_idx
            ]

            right = obs_idx[
                obs_idx > col_idx
            ]

            # gap inside smile
            if len(left) > 0 and len(right) > 0:

                pred = pred_poly

            # gap at smile edge
            else:

                # first wing
                if len(left) == 0:

                    n1 = obs_idx[0]
                    n2 = obs_idx[1]

                # last wing
                else:

                    n1 = obs_idx[-2]
                    n2 = obs_idx[-1]

                # local wing slope
                slope = (
                    iv[n2] - iv[n1]
                ) / (
                    lm[n2] - lm[n1]
                )

                # linear wing estimate
                pred_lin = (
                    iv[n1]
                    +
                    slope
                    *
                    (
                        lm[col_idx]
                        -
                        lm[n1]
                    )
                )

                # final edge blend
                # 9% smile
                # 91% linear wing
                pred = (
                    0.09 * pred_poly
                    +
                    0.91 * pred_lin
                )

            # keep IV positive
            blend09_df.at[
                i,
                opt_cols[col_idx]
            ] = max(
                pred,
                0.001
            )
```

```python
# wide surface -> long format
# one row per timestamp-option pair

blend09_long = blend09_df.melt(
    id_vars=["datetime"],
    value_vars=ce_cols + pe_cols,
    var_name="option",
    value_name="pred_iv"
)
```

```python
# keep track of observed and missing IVs

pred_blend09 = test_df.melt(
    id_vars=["datetime"],
    value_vars=ce_cols + pe_cols,
    var_name="option",
    value_name="iv"
)
```

```python
## keep missing IV entries only

pred_blend09 = pred_blend09[
    pred_blend09["iv"].isna()
].copy()
```

```python
# attach reconstructed IV values
# match each missing location with its prediction

pred_blend09 = pred_blend09.merge(
    blend09_long,
    on=["datetime", "option"],
    how="left"
)
```

```python
# string date -> datetime object
# easier to format later

pred_blend09["datetime"] = pd.to_datetime(
    pred_blend09["datetime"],
    dayfirst=True
)
```

```python
# create final submission id in required format

pred_blend09["id"] = (
    pred_blend09["datetime"]
        .dt.strftime("%d-%m-%Y %H:%M")
    + "||"
    + pred_blend09["option"]
)
```

```python
## creating final submission :)

sub_blend09 = pred_blend09[
    ["id", "pred_iv"]
].rename(
    columns={
        "pred_iv": "value"
    }
)
```
